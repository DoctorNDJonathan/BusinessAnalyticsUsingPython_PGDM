## 2 · Variables and data types (20 min)

A **variable** is a labelled box that holds a value. The `=` sign means *"put the value on the right into the box on the left"* — it is **not** "equals" in the maths sense.

| Type | Business example | Python |
|---|---|---|
| `int` (whole number) | cups sold | `148` |
| `float` (decimal) | price, revenue | `120.0` |
| `str` (text) | café name, item | `"Namma Brew"` |
| `bool` (True/False) | open on Sunday? | `True` |

````python
cafe_name      = "Namma Brew"
cups_sold      = 148
price_per_cup  = 120.0
open_on_sunday = True

print(type(cafe_name), type(cups_sold), type(price_per_cup), type(open_on_sunday))
````

````python
revenue = cups_sold * price_per_cup
print(revenue)
print(type(revenue))   # PREDICT: int or float? Why?
````

**Naming rules (and good habits)**
- Letters, digits and `_` only; cannot start with a digit → `2024_sales` ❌ `sales_2024` ✅
- **Case-sensitive**: `Revenue` and `revenue` are two different boxes
- Use `snake_case` names that say what the value *means*: `daily_cups`, not `x1`

### 2.1 Converting between types
`int()`, `float()`, `str()` convert values. This is how we repair the Hook problem.

````python
price = float(price_from_file)   # "120" -> 120.0
qty   = int(qty_from_file)       # "3"   -> 3
print("Bill:", price * qty)
````

````python
# PREDICT: will this work?
int("120.50")
````

````python
# Fix: go text -> float first, then float -> int if you really need a whole number
print(float("120.50"))
print(int(float("120.50")))   # PREDICT: 120 or 121?
````

### 2.2 Arithmetic that managers actually use

````python
cups = 148
box_size = 12      # take-away cups are packed in boxes of 12

print(cups / box_size)    # true division  -> always a float
print(cups // box_size)   # floor division -> full boxes
print(cups % box_size)    # remainder      -> loose cups left over
print(2 ** 10)            # power
````

Computers store most decimals as close approximations, not exact values. For reports, **round at the end**. For real accounting systems, Python offers the `decimal` module for exact money arithmetic.

### 3.1 Lists — an ordered row of values
Cups sold Monday → Sunday:

````python
days       = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
daily_cups = [148, 132, 160, 95, 170, 210, 198]

print(daily_cups[0])     # PREDICT: which day?
print(daily_cups[-1])    # PREDICT: which day?
print(daily_cups[1:3])   # PREDICT: how many values?
````

⚠️ **Python counts from 0.** And a slice `[1:3]` includes position 1 but **stops before** position 3.

````python
print("Days recorded :", len(daily_cups))
print("Total cups    :", sum(daily_cups))
print("Best day cups :", max(daily_cups))
print("Worst day cups:", min(daily_cups))
print("Average       :", sum(daily_cups) / len(daily_cups))
````

````python
best_position = daily_cups.index(max(daily_cups))
print("Best day was", days[best_position])
````

### 3.2 ⚠️ Trap: two names, one list

````python
week1 = [148, 132, 160]
backup = week1          # "make a backup"... or did we?
backup.append(999)      # a typo added to the backup

print("backup:", backup)
print("week1 :", week1)   # PREDICT before running
````

````python
week1 = [148, 132, 160]
backup = week1.copy()   # a genuine, separate copy
backup.append(999)
print("backup:", backup)
print("week1 :", week1)
````

`backup = week1` does **not** copy — it gives the *same* list a second name. You will meet this idea again with pandas tables.

### 3.3 Dictionaries — look up a value by its name (key)

## 4 · Control flow: making decisions and repeating work (15 min)

### 4.1 if / elif / else — the café's loyalty discount

| Bill amount | Discount |
|---|---|
| ₹2,000 and above | 15% |
| ₹1,000 – 1,999 | 10% |
| ₹500 – 999 | 5% |
| below ₹500 | none |

Python checks conditions **from top to bottom** and stops at the **first** one that is True. Indentation (4 spaces) shows which lines belong to which condition.

````python
bill = 2500

if bill >= 2000:
    discount = 0.15
elif bill >= 1000:
    discount = 0.10
elif bill >= 500:
    discount = 0.05
else:
    discount = 0.0

print(f"Bill ₹{bill} → discount {discount:.0%}")
````

### ⚠️ Trap: same conditions, different order
A new intern rewrote the rule. **Predict** the discount on a ₹2,500 bill before running.

````python
bill = 2500

if bill >= 500:
    discount = 0.05
elif bill >= 1000:
    discount = 0.10
elif bill >= 2000:
    discount = 0.15
else:
    discount = 0.0

print(f"Bill ₹{bill} → discount {discount:.0%}")   # Is this what the policy says?
````

No error — but the café's best customers get the *smallest* discount. **Order matters**: test the most demanding condition first.

Also remember: `=` **assigns**, `==` **compares**.

````python
target = 150
print(160 == target, 150 == target, 150 >= target, 150 != target)
````

### 4.2 for loops — do the same thing for every item
Which days missed the target of 150 cups? `zip()` walks through two lists side by side.

````python
target = 150

for day, cups in zip(days, daily_cups):
    if cups < target:
        print(f"{day}: {cups} cups — below target by {target - cups}")
````

**The accumulator pattern** — start a running total at zero, add to it inside the loop. You will use this everywhere.

````python
menu = {"Filter Coffee": 40, "Cappuccino": 120, "Masala Chai": 30, "Bun Maska": 50}
order = [("Cappuccino", 2), ("Filter Coffee", 3), ("Bun Maska", 2)]   # (item, quantity)

subtotal = 0
for item, qty in order:
    line_total = menu[item] * qty
    subtotal = subtotal + line_total
    print(f"{item:<15} x{qty}  ₹{line_total}")

print(f"Subtotal: ₹{subtotal}")
````

### 4.3 while loops — repeat until a condition changes
The café has 5,000 g of coffee beans. Each cup uses 18 g. Reorder when stock falls to 500 g. How many cups before we must reorder?

````python
beans_g = 5000
grams_per_cup = 18
reorder_level = 500
cups_made = 0

while beans_g - grams_per_cup >= reorder_level:
    beans_g = beans_g - grams_per_cup
    cups_made = cups_made + 1

print(f"Cups before reorder: {cups_made}; beans left: {beans_g} g")
````

⚠️ If the condition can never become False, the loop runs forever. Use the **Stop** button and check that something inside the loop changes the condition.

## 5 · Functions — a preview (7 min)

A **function** packages a calculation so you can reuse it with different inputs. We go deeper next session.

````python
def bill_with_gst(subtotal, gst_rate=0.05):
    """Return the bill including GST, rounded to 2 decimals."""
    return round(subtotal * (1 + gst_rate), 2)

print(bill_with_gst(460))
print(bill_with_gst(1768))
print(bill_with_gst(1000, gst_rate=0.18))   # a different rate, same function
````

### ⚠️ Trap: `print` is not `return`

````python
def bill_with_gst_v2(subtotal, gst_rate=0.05):
    print(round(subtotal * (1 + gst_rate), 2))   # shows the number... but gives nothing back

total = bill_with_gst_v2(460)
print("Stored value:", total)    # PREDICT
````

`print` shows a value to a human. `return` hands it back to the program so it can be stored and used. A function with no `return` gives back `None`.

### 5.1 Putting it all together — one complete bill
(GST rate here is an assumption for the exercise.)

````python
menu = {"Filter Coffee": 40, "Cappuccino": 120, "Masala Chai": 30, "Bun Maska": 50}
office_order = [("Cappuccino", 10), ("Filter Coffee", 12), ("Bun Maska", 8)]

# 1. Subtotal (accumulator)
subtotal = 0
for item, qty in office_order:
    subtotal += menu[item] * qty          # += is shorthand for subtotal = subtotal + ...

# 2. Discount (if / elif / else)
if subtotal >= 2000:
    discount = 0.15
elif subtotal >= 1000:
    discount = 0.10
elif subtotal >= 500:
    discount = 0.05
else:
    discount = 0.0
after_discount = subtotal * (1 - discount)

# 3. Tax (function)
final = bill_with_gst(after_discount)

print(f"Subtotal       ₹{subtotal:,.2f}")
print(f"Discount       {discount:.0%}")
print(f"After discount ₹{after_discount:,.2f}")
print(f"Payable (GST)  ₹{final:,.2f}")
````
